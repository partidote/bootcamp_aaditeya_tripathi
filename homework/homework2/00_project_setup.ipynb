# Environment & Config Check
This notebook verifies Python environment, .env loading, and basic NumPy operations.

```python
import sys, os
import numpy as np
from pathlib import Path
from dotenv import load_dotenv

print("Python version:", sys.version)
print("Interpreter path:", sys.executable)
```

```python
load_dotenv()
api_key = os.getenv("API_KEY")
data_dir = os.getenv("DATA_DIR", "./data")

print("API_KEY present:", api_key is not None)
print("DATA_DIR from env:", data_dir)

Path(data_dir).mkdir(parents=True, exist_ok=True)
print("Ensured data directory exists.")
```

```python
arr = np.array([1, 2, 3, 4, 5])
print("Array:", arr)
print("Mean:", arr.mean())
print("Vectorized * 10:", arr * 10)
```
